```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
model_id = 'Falconsai/text_summarization'  #repo of the model in huggingface
tokenizer = AutoTokenizer.from_pretrained(model_id)  #extraction of the necessary tokenizer
model = AutoModelForSeq2SeqLM.from_pretrained(model_id)  #extraction of the required model
inputs = tokenizer(article,return_tensors="pt", truncation=True).input_ids   #conversion of the article into tokens with the help of tokenizer
#with the help of truncation, we can limit the number of input tokens within the threshold of number of tokens that can be accepted by the model.
output = model.generate(inputs,max_new_tokens = 500, do_sample=False)   #models generated summary 
summary = tokenizer.decode(output[0],skip_special_tokens=True)
print(summary)
```

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
article = 'Dance is an art form, consisting of sequences of body movements with aesthetic and often symbolic value, either improvised or purposefully selected. Dance can be categorized and described by its choreography, by its repertoire of movements or by its historical period or place of origin. Dance is typically performed with musical accompaniment, and sometimes with the dancer simultaneously using a musical instrument themselves. Two common types of group dance are theatrical and participatory dance. Both types of dance may have special functions, whether social, ceremonial, competitive, erotic, martial, sacred or liturgical. Dance is not solely restricted to performance, as dance is used as a form of exercise and occasionally training for other sports and activities. Dance performances and dancing competitions are found across the world exhibiting various different styles and standards. Dance may also be participated in alone as a form of exercise or self expression. Dancing is common human behaviour, and does not necessarily require specific choreography.'
model_id = 'facebook/bart-large-cnn'  #repo of the model in huggingface
tokenizer = AutoTokenizer.from_pretrained(model_id)  #extraction of the necessary tokenizer
model = AutoModelForSeq2SeqLM.from_pretrained(model_id)  #extraction of the required model
inputs = tokenizer(article,return_tensors="pt", truncation=True)   #conversion of the article into tokens with the help of tokenizer
#with the help of truncation, we can limit the number of input tokens within the threshold of number of tokens that can be accepted by the model.
output = model.generate(inputs['input_ids'],max_length = 60, min_length = 10)   #models generated summary 
summary = tokenizer.decode(output[0],skip_special_tokens=True)
print(summary)
```

Importing the necessary tokenizer and the model for the language translation

```python
from transformers import MBartForConditionalGeneration, MBart50TokenizerFast
model = MBartForConditionalGeneration.from_pretrained("facebook/mbart-large-50-many-to-many-mmt")
tokenizer = MBart50TokenizerFast.from_pretrained("facebook/mbart-large-50-many-to-many-mmt")
```

checking whether the model can catch the slang or no

```python
nepali_text = 'बबाल छ!'
tokenizer.src_lang = 'ne_NP'  #setting the source language to our tokenizer
trns_lang = 'en_XX'     #the targeted language
encoded_txt = tokenizer(nepali_text, return_tensors="pt")  #converting the given text into tokens for the model
generated_tokens = model.generate(**encoded_txt,forced_bos_token_id=tokenizer.lang_code_to_id[trns_lang])
tokenizer.batch_decode(generated_tokens, skip_special_tokens=True)[0]
```

As we can see, the model hasnot been trained on the slang


```python
nepali_text = 'I love you.'
tokenizer.src_lang = 'en_XX'   #setting the source language to our tokenizer
trns_lang =  'fr_XX'   #the targeted language
encoded_txt = tokenizer(nepali_text, return_tensors="pt")  #converting the given text into tokens for the model
generated_tokens = model.generate(**encoded_txt,forced_bos_token_id=tokenizer.lang_code_to_id[trns_lang])
tokenizer.batch_decode(generated_tokens, skip_special_tokens=True)[0]
```

Loading the auto tokenizer


```python
tokenizer = AutoTokenizer.from_pretrained("distilbert/distilgpt2")
```

```python
def apply_tokenizer(data):
    return tokenizer([' '.join(x) for x in data['answers.text']]) 

```

applying the tokenizer function on all the examples of training data

```python
tokenized_train_data = train_data.map(apply_tokenizer,
                                     batched = True,
                                     num_proc = 4,
                                     remove_columns = train_data.column_names)
```
