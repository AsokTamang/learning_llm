Text summarization

```python
article = "By default, React Compiler will memoize your code based on its analysis and heuristics. In most cases, this memoization will be as precise, or moreso, than what you may have written. However, in some cases developers may need more control over memoization. The useMemo and useCallback hooks can continue to be used with React Compiler as an escape hatch to provide control over which values are memoized. A common use-case for this is if a memoized value is used as an effect dependency, in order to ensure that an effect does not fire repeatedly even when its dependencies do not meaningfully change. For new code, we recommend relying on the compiler for memoization and using useMemo/useCallback where needed to achieve precise control. For existing code, we recommend either leaving existing memoization in place (removing it can change compilation output) or carefully testing before removing the memoization."
```

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
model_id = 'Falconsai/text_summarization'  #repo of the model in huggingface
tokenizer = AutoTokenizer.from_pretrained(model_id)  #extraction of the necessary tokenizer
model = AutoModelForSeq2SeqLM.from_pretrained(model_id)  #extraction of the required model
inputs = tokenizer(article,return_tensors="pt", truncation=True).input_ids   #conversion of the article into tokens with the help of tokenizer
output = model.generate(inputs,max_new_tokens = 500, do_sample=False)   #models generated summary 
summary = tokenizer.decode(output[0],skip_special_tokens=True)
print(summary)
```

Use of openai large language model

```python

import os
from huggingface_hub import InferenceClient
client = InferenceClient()
completion = client.chat.completions.create(
    model = "openai/gpt-oss-120b",
    messages = [
        
        {
            "role": "user",
            "content": "Explain me the term cosmos in simple language"
        }
    ]
)
print(completion.choices[0].message)
```

Use of the model to fill the masked word

```python
unmasker = pipeline('fill-mask',model = 'bert-base-cased')
unmasker('Hi, I want to [MASK] with you.',top_k = 2)  #here we are using top_k as 2 inorder to provide only 2 outputs
```

Name entity recognition model

```python
ner_model = pipeline("ner", aggregation_strategy="simple")  #here with the help of aggregation_strategy function we are able to group the two words asok tamang in the same word category 
ner_model('I am Asok Tamang and she is Sylvia Grey,I am from Nepal and She is from Canada.')  #so the lettter case of the name of the person as well as the place or country matters for name entity recognition model
```
