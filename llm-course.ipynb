Fine tuning a text generation model

```python
from datasets import load_dataset

train_data = load_dataset("dany0407/eli5_category", split="train[:5000]")
train_data = train_data.train_test_split(test_size=0.2)  #splitting the data into train and test based on train test split
```

```python
train_data = train_data.flatten()
```

Loading the auto tokenizer


```python
tokenizer = AutoTokenizer.from_pretrained("distilbert/distilgpt2")
```

```python
def apply_tokenizer(data):
    return tokenizer([' '.join(x) for x in data['answers.text']]) 

```

applying the tokenizer function on all the examples of training data

```python
tokenized_train_data = train_data.map(apply_tokenizer,
                                     batched = True,
                                     num_proc = 4,
                                     remove_columns = train_data['train'].column_names)
```

Grouping the tokens into the limited chunks and this below approach does save the data, even though the number of tokens are shorter than the block-size.

```python
block_size = 128
def group_into_chunks(x):
    concatated_data = {k: sum(x[k],[]) for k in x.keys()}  #making a dictionary based on the key and the tokens as the values
    total_length = len(concatated_data[list(x.keys())[0]])  #here we are only counting the length of the input_ids excluding that of the attention mask
    if total_length >= block_size:  #if the total length exceeds the block size then we must limit the total length within the block size chunks
        total_length = (total_length // block_size) * block_size
    result = {
        k:[t[i:i+block_size] for i in range(0,total_length,block_size)] for k,t in concatated_data.items() 
    }    #here we are storing the final range of tokens within the limit of block_size and storing the tokens based on the key
    result['labels'] = result['input_ids'].copy()  #then finalizing the result with the key labels only escaping the other key such as the paddings and so on
    return result
        
```

Grouping the training data into chunks

```python
grouped_train_data = tokenized_train_data.map(group_into_chunks,batched = True , num_proc = 4)
```

The below approach doesnot save the datas as the tokens which are shorter than the block-size are dropped.

Preparing data-collator for casual language modeling

```python
from transformers import DataCollatorForLanguageModeling
tokenizer.pad_token = tokenizer.eos_token  #setting the pad token of tokenizer to end of sequence token
data_collator = DataCollatorForLanguageModeling(tokenizer = tokenizer, mlm = False)
```

Importing the necessary model

```python
from transformers import AutoModelForCausalLM, TrainingArguments, Trainer
model = AutoModelForCausalLM.from_pretrained("distilbert/distilgpt2")
```

Training the model

```python
training_args = TrainingArguments(
    output_dir = 'trained_clm',
    eval_strategy="epoch",
    learning_rate=2e-5,
    weight_decay=0.01,
    push_to_hub=False
)
trainer = Trainer(
    model = model,
    args = training_args,
    train_dataset = grouped_train_data['train'],
    eval_dataset = grouped_train_data['test'],
    data_collator = data_collator,
    processing_class = tokenizer
)
trainer.train()
```

Fine-tuning a text classification model

Loading the dataset

```python
from datasets import load_dataset
imdb = load_dataset("stanfordnlp/imdb")
```

```python
imdb['test'][0]
```

Function to tokenize the inputs

```python
tokenizer = AutoTokenizer.from_pretrained("distilbert/distilbert-base-uncased")
def preprocess_function(examples):
    return tokenizer(examples['text'],truncation = True)
    
```

```python
tokenized_imdb = imdb.map(preprocess_function, batched = True)  #tokenizing the input data

```

```python
from transformers import DataCollatorWithPadding
data_collator = DataCollatorWithPadding(tokenizer = tokenizer)   
```

Evaluation of model's performance

```python
!pip install evaluate 
import evaluate
accuracy = evaluate.load('accuracy')
```

Function to compute the metric


```python
import numpy as np
def compute_metrics(eval_pred):
    predictions , labels = eval_pred   #here we are extracting the prediction of our model and the actual label of the data
    predictions = np.argmax(predictions,axis = 1)   #as we have the list of probabilities , we choose the index having the highest probability value
    return accuracy.compute(predictions = predictions, references = labels)  #calculation of how accurate our model is
    
    
```

Assigning the id with their corresponding labels and vice-versa

```python
id2label = {0: "NEGATIVE", 1: "POSITIVE"}
label2id = {"NEGATIVE": 0, "POSITIVE": 1}
```

Defining our classification model

```python
from transformers import AutoModelForSequenceClassification, TrainingArguments, Trainer
model = AutoModelForSequenceClassification.from_pretrained("distilbert/distilbert-base-uncased", num_labels=2, id2label=id2label, label2id=label2id)
```

Training the model

```python
training_args = TrainingArguments(
    output_dir="my_classification_model",
    learning_rate=2e-5,
    per_device_train_batch_size=16,
    per_device_eval_batch_size=16,
    num_train_epochs=2,
    weight_decay=0.01,
    eval_strategy="epoch",
    save_strategy="epoch",
    load_best_model_at_end=True,
    push_to_hub=True,
)
#we must tokenize the training data before passing it to the model for training 
trainer = Trainer(
    model = model,
    args = training_args,
    train_dataset = tokenized_imdb['train'],
    eval_dataset = tokenized_imdb['test'],
    processing_class = tokenizer,
    data_collator = data_collator,
    compute_metrics = compute_metrics
    )
trainer.train()

```
